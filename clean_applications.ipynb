# Clean filters in the applications lists

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('20251005_application_list.xlsx')
```

```python
df.columns
```

```python
def unique_entries(series):
    return pd.unique(series.str.split(', ',expand=True).values.ravel('K'))
```

## Unique types

```python
unique_entries(df['Type'])
```

```python
df['Type']=df['Type'].replace(np.nan,None)
```

```python
unique_entries(df['Type'])
```

## Target audiences

```python
unique_entries(df['Target audiences'])
```

```python
df['Target audiences']=df['Target audiences'].replace(np.nan,None)
```

```python
unique_entries(df['Target audiences'])
```

## Scale

```python
unique_entries(df['Scale'])
```

```python
df['Scale']=df['Scale'].replace(np.nan,None)
```

```python
df['Scale']=df['Scale'].str.replace('Country','National')\
    .str.replace('Cities','City')
```

```python
unique_entries(df['Scale'])
```

## Geographies

```python
df['Geographies']=df['Geographies'].replace(np.nan,None)
```

```python
df['Geographies']=df['Geographies']\
    .str.replace('Columbia','Colombia')\
    .str.replace('EU','European Union')\
    .str.replace('New Zealand','New Zeeland')\
    .str.replace('Russian Federation','RPlaceholder')\
    .str.replace('Russia','RPlaceholder')\
    .str.replace('RPlaceholder','Russian Federation')\
    .str.replace('The Gambia','Gambia')\
    .str.replace('Vietnam','Viet Nam')\
    .str.replace('UAE','United Arab Emirates')\
    .str.replace('UK','United Kingdom')\
    .str.replace('USA','United States')\
    .str.replace('United States of America','United States')
```

```python
unique_entries(df['Geographies'])
```

## Main topic

```python
unique_entries(df['Main topic'])
```

```python
df['Main topic']=df['Main topic'].replace(np.nan,None)
```

```python
unique_entries(df['Main topic'])
```

```python
df.to_csv('20251006_application_list.csv',index=None)
```
